```python
import os
import sys
import importlib.util
import nemo.collections.asr as nemo_asr
from huggingface_hub import hf_hub_download

REPO = "qvac/TranscribePsy-250M-EU"
model = nemo_asr.models.ASRModel.from_pretrained(REPO).eval()

# Download and dynamically import inference.py
inference_script_path = hf_hub_download(repo_id=REPO, filename="inference.py")
spec = importlib.util.spec_from_file_location("inference", inference_script_path)
inference = importlib.util.module_from_spec(spec)
sys.modules["inference"] = inference
spec.loader.exec_module(inference)

# Download the adapters file
adapters_path = hf_hub_download(repo_id=REPO, filename="adapters.safetensors")

# Attach adapters for non-English languages (including Lithuanian)
inference.attach_adapters(
    model, 
    adapters_path, 
    device=model.device, 
    dtype=next(model.parameters()).dtype
)

# Transcribe Lithuanian audio
print(model.transcribe(["test_mono.wav"])[0].text)
```
